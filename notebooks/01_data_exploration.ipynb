# Exploração de Dados - Datathon Passos Mágicos

## Objetivo
Exploração inicial dos dados de desenvolvimento educacional (2022-2024) para compreender estrutura, qualidade e características.

## Processo
1. Carregar dados dos sheets Excel
2. Avaliar estrutura e tipos de dados
3. Identificar valores faltantes
4. Gerar estatísticas descritivas
5. Documentar achados iniciais

```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configuração inicial
project_root = Path.cwd().parent
DATA_FILE = project_root / "BASE DE DADOS PEDE 2024 - DATATHON.xlsx"

# Paleta de cores
COLORS = {
    'primary': '#2c3e50',
    'secondary': '#34495e',
    'accent': '#3498db',
    'warning': '#e74c3c',
    'success': '#27ae60',
    'neutral': '#95a5a6',
    'background': '#ecf0f1',
    'text': '#2c3e50'
}

# Configuração dos gráficos
PLOT_CONFIG = {
    'figure.facecolor': 'white',
    'axes.facecolor': '#f8f9fa',
    'axes.edgecolor': COLORS['primary'],
    'axes.labelcolor': COLORS['text'],
    'text.color': COLORS['text'],
    'xtick.color': COLORS['text'],
    'ytick.color': COLORS['text'],
    'grid.color': '#d5d8dc',
    'grid.alpha': 0.3,
    'font.size': 10,
    'axes.titlesize': 12,
    'axes.labelsize': 10,
    'figure.titlesize': 14
}

plt.rcParams.update(PLOT_CONFIG)
sns.set_palette([COLORS['primary'], COLORS['accent'], COLORS['secondary'], 
                 COLORS['success'], COLORS['warning'], COLORS['neutral']])

# Configurações de exibição
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

print(f"Local do arquivo de dados: {DATA_FILE}")
print(f"Arquivo existe: {DATA_FILE.exists()}")
```

## 1. Carregamento de Dados

Carregando arquivo Excel com três sheets (2022, 2023, 2024).

```python
# Carregar todos os sheets
data_dict = {}
excel_file = pd.ExcelFile(DATA_FILE)

print(f"Número de sheets: {len(excel_file.sheet_names)}")
print(f"Nomes dos sheets: {excel_file.sheet_names}")
print("\n" + "="*80)

# Exibir dimensões de cada sheet
for sheet in excel_file.sheet_names:
    df = pd.read_excel(excel_file, sheet_name=sheet)
    data_dict[sheet] = df
    print(f"\n{sheet}: {df.shape[0]} linhas × {df.shape[1]} colunas")
```

## 2. Estrutura Inicial do Dataset

Inspeção das colunas, tipos de dados e primeiras linhas do primeiro ano disponível.

```python
# Selecionar o primeiro sheet para análise detalhada
first_year = list(data_dict.keys())[0]
df_sample = data_dict[first_year]

print(f"Analisando estrutura de: {first_year}")
print("\nNomes das colunas:")
print(df_sample.columns.tolist())
```

```python
# Tipos de dados e contagens não-nulas
print("\nInformações dos Dados:")
print(df_sample.info())
```

```python
# Visualizar primeiras linhas
print(f"\nPrimeiras 10 linhas de {first_year}:")
df_sample.head(10)
```

## 3. Avaliação da Qualidade dos Dados

Identificando valores faltantes, duplicatas e possíveis problemas.

```python
# Análise de valores faltantes por ano
for year, df in data_dict.items():
    print(f"\n{'='*80}")
    print(f"Valores Faltantes - {year}")
    print(f"{'='*80}")
    
    missing_counts = df.isnull().sum()
    missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    missing_summary = pd.DataFrame({
        'Missing_Count': missing_counts,
        'Missing_Pct': missing_pct
    })
    
    missing_summary = missing_summary[missing_summary['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)
    
    if len(missing_summary) > 0:
        print(missing_summary)
    else:
        print("Nenhum valor faltante detectado.")
```

```python
# Verificar duplicatas
print("\nLinhas Duplicadas:")
print("="*80)
for year, df in data_dict.items():
    duplicates = df.duplicated().sum()
    print(f"{year}: {duplicates} linhas duplicadas")
```

## 4. Estatísticas Descritivas

Resumo estatístico das colunas numéricas.

```python
# Estatísticas descritivas para o primeiro ano
print(f"\nEstatísticas Descritivas - {first_year}")
stats = df_sample.describe().T
stats['dtype'] = df_sample.dtypes
stats['missing'] = df_sample.isnull().sum()
stats['missing_pct'] = (df_sample.isnull().sum() / len(df_sample) * 100).round(2)
print(stats)
```

```python
# Identificar colunas numéricas e indicadores
print("\nAnálise de Colunas por Ano:")
print("="*80)
for year, df in data_dict.items():
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    print(f"\n{year}: {len(numeric_cols)} colunas numéricas")
    print(f"Numéricas: {numeric_cols[:10]}..." if len(numeric_cols) > 10 else f"Numéricas: {numeric_cols}")
```

## 5. Observações Iniciais

Documentar achados para orientar os próximos passos de limpeza e análise.

```python
# Resumo das observações
observations = []

for year, df in data_dict.items():
    obs = {
        'Ano': year,
        'Total_Linhas': df.shape[0],
        'Total_Colunas': df.shape[1],
        'Valores_Faltantes': df.isnull().sum().sum(),
        'Linhas_Duplicadas': df.duplicated().sum(),
        'Colunas_Numericas': len(df.select_dtypes(include=[np.number]).columns)
    }
    observations.append(obs)

summary_df = pd.DataFrame(observations)
print("\nResumo de Dados por Anos:")
print(summary_df)
```
